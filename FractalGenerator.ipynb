```python
import numpy as np
from PIL import Image
import cmath

# 1. Define the function and its derivative
def f(z):
    return z**3 - 1

def f_prime(z):
    return 3 * z**2

# 2. Define the roots of z^3 - 1 = 0
# These are the cube roots of unity
roots = [
    1 + 0j,  # 1
    cmath.exp(2j * cmath.pi / 3),  # -0.5 + i*sqrt(3)/2
    cmath.exp(4j * cmath.pi / 3)   # -0.5 - i*sqrt(3)/2
]

# 3. Parameters for the fractal generation
WIDTH, HEIGHT = 800, 800  # Image resolution
MAX_ITERATIONS = 50       # Maximum iterations for Newton's method
TOLERANCE = 1e-6          # Convergence tolerance

# Complex plane boundaries
X_MIN, X_MAX = -1.5, 1.5
Y_MIN, Y_MAX = -1.5, 1.5

# 4. Custom colors for the roots and divergence
# These are placeholder university colors. Replace with your actual RGB values!
# Example: For a university with blue and gold, you might use:
# ROOT_COLORS = [(0, 0, 139), (255, 215, 0), (128, 0, 128)] # DarkBlue, Gold, Purple
# DIVERGENCE_COLOR = (0, 0, 0) # Black for points that don't converge

ROOT_COLORS = [
    (255, 82, 0),   # ECU Orange for root 1
    (35, 31, 32),   # black
    (33, 156, 215)    # Blue for root 3
]
DIVERGENCE_COLOR = (0, 0, 0) # Black for points that don't converge

# 5. Create an image buffer
img_data = np.zeros((HEIGHT, WIDTH, 3), dtype=np.uint8)

# 6. Iterate over each pixel in the image grid
for py in range(HEIGHT):
    for px in range(WIDTH):
        # Map pixel coordinates to complex plane coordinates
        x = X_MIN + (px / WIDTH) * (X_MAX - X_MIN)
        y = Y_MIN + (py / HEIGHT) * (Y_MAX - Y_MIN)
        z = complex(x, y)

        converged_to_root = -1
        iterations = 0

        # Apply Newton's method
        for i in range(MAX_ITERATIONS):
            if abs(f_prime(z)) < 1e-10: # Avoid division by zero if derivative is too small
                break
            z_next = z - f(z) / f_prime(z)

            # Check for convergence to any root
            for root_idx, root in enumerate(roots):
                if abs(z_next - root) < TOLERANCE:
                    converged_to_root = root_idx
                    break

            if converged_to_root != -1:
                break

            z = z_next
            iterations = i + 1

        # Assign color based on convergence
        if converged_to_root != -1:
            # Interpolate color based on iterations for a smoother look
            base_color = ROOT_COLORS[converged_to_root]
            # A simple iteration-based shading: darker for faster convergence
            shading_factor = 1 - (iterations / MAX_ITERATIONS) * 0.7 # Adjust 0.7 for intensity

            r = int(base_color[0] * shading_factor)
            g = int(base_color[1] * shading_factor)
            s_b = int(base_color[2] * shading_factor)
            img_data[py, px] = (r, g, s_b)
        else:
            img_data[py, px] = DIVERGENCE_COLOR # Points that did not converge

# 7. Create and save the image
img = Image.fromarray(img_data, 'RGB')
output_filename = 'newton_fractal_z3_minus_1.png'
img.save(output_filename)
print(f"Fractal image saved as {output_filename}")
```
