## Fisher Yates Shuffle

```python
import random

st = "ABCDE"

for i in range(len(st)-1,0,-1):
  j = random.randint(0,i-1)
  st = st[:j] +st[i] + st[j+1:i] + st[j] + st[i+1:]

print(st)
```

The above code does not work.  Because I rely on j being less than i, it will never leave the first char in the first position.  instead I needed to convert the string to a list so that I could swap the chars in place without reconstructing the rest of the string.

```python


def shuffle(st):
  stl = list(st)
  for i in range(len(st)-1,0,-1):
    j = random.randint(0,i)
    stl[i], stl[j] = stl[j], stl[i]
  st = "".join(stl)
  return st

shuffle("ABCDE")
```

```python
l = []

for i in range(1000):
  l.append(shuffle("ABCDE")[0])

```

Below I confirm to myself that I did shuffle the first digit.

```python
import pandas as pd


s = pd.Series(l)
counts = s.value_counts()

counts.sort_index().plot.bar()
```
